###Eliminaciones realizadas:

1. ``` weight_kg ``` No es adecuado imputar peso con una media (o mediana) simple porque el peso depende de la especie, raza, tamaño y edad.

2. ``` household_size ``` Depende del tipo de vivienda y ubicación del adoptante; imputar un promedio o mediana distorsionaría el entorno real.

3. ``` microchipped ``` Siendo binaria, rellenarla sería asumir un estado sin evidencia.

````python
df = df.dropna(subset = ["weight_kg"])
````

````python
df = df.dropna(subset = ["household_size"])
````

````python
df = df.dropna(subset = ["microchipped"])
````

````python
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(df.isnull(), cbar = False)
plt.title("Mapa de valores faltantes")
plt.show()
````

###Imputación de datos faltantes (vacíos):

1. ``` pet_name ``` Se imputaron con "desconocido" ya que el nombre de la mascota es un dato trivial (al menos en este análisis)

2. ``` owner_satisfaction_rating ``` Se imputaron con la media de los datos al considerar favorable esta medida por mantener la distribución de los datos

3. ``` annual_grooming_cost_usd ``` Se imputaron con la media de los datos POR ESPECIE al considerar favorable esta medida por sobre la mediana

````python
df["pet_name"] = df["pet_name"].fillna("desconocido")
````

````python
promedio = float(df["owner_satisfaction_rating"].mean())
df["owner_satisfaction_rating"] = df["owner_satisfaction_rating"].fillna(promedio)
````

````python
df['annual_grooming_cost_usd'] = df.groupby('species')['annual_grooming_cost_usd'].transform(lambda x: x.fillna(x.median()))
````

````python
import seaborn as sns
import matplotlib.pyplot as plt
sns.heatmap(df.isnull(), cbar = False)
plt.title("Mapa de valores faltantes")
plt.show()
````

````python
#Anualización del costo de la comida (Human made)
df['annual_food_cost_usd'] = df['monthly_food_cost_usd'] * 12

cost_columns = [
    'annual_food_cost_usd',
    'annual_vet_cost_usd',
    'annual_grooming_cost_usd',
    'annual_toys_accessories_cost_usd',
    'annual_training_cost_usd',
    'annual_insurance_cost_usd',
]
cost_labels = [
    'Comida',
    'Veterinaria',
    'Peluquería / Higiene',
    'Juguetes y Accesorios',
    'Entrenamiento',
    'Seguro Médico',
]
# Código para promedios por especie (Human made)
gastos_promedio = df.groupby('species')[cost_columns].mean()
gastos_promedio.columns = cost_labels
# Paleta de colores fija para que cada rubro tenga el mismo color en todos los pasteles
colores = plt.cm.Set2.colors[: len(cost_labels)]
# Crear la cuadrícula integrada: 2 filas x 4 columnas
fig, axes = plt.subplots(nrows=2, ncols=4, figsize=(16, 8))
axes_flat = axes.flatten()  # Aplanar la matriz 2D a una lista de 8 espacios
# Función para no mostrar porcentajes menores al 4% (evita que se encimen textos de rubros casi en 0)
formato_pct = lambda p: f'{p:.0f}%' if p >= 4 else ''
# Graficar cada especie en su respectivo cuadrante
for i, especie in enumerate(gastos_promedio.index):
  ax = axes_flat[i]
  valores = gastos_promedio.loc[especie]

  wedges, _, _ = ax.pie(
      valores,
      autopct=formato_pct,
      startangle=140,
      colors=colores,
      pctdistance=0.75,
      wedgeprops=dict(width=0.85, edgecolor='white', linewidth=1.5),
  )
  ax.set_title(especie, fontsize=13, fontweight='bold', pad=5)
# Usar el espacio restante (índice 7) para colocar la Leyenda General Compartida
ax_legend = axes_flat[7]
ax_legend.axis('off')  # Oculta los ejes del recuadro
ax_legend.legend(
    wedges,
    cost_labels,
    title='Rubros de Gasto',
    title_fontsize=12,
    fontsize=11,
    loc='center',
    frameon=True,
    edgecolor='gray',
)
plt.suptitle(
    'Distribución del Gasto Anual por Especie',
    fontsize=16,
    fontweight='bold',
    y=0.98,
)
plt.tight_layout()
plt.show()
````

Conclusiones generales:
1. La comida y la veterinaria se llevan entre el 64% y el 80% del gasto anual
2. Exclusividad del entrenamiento en los perros
3. Peso porcentual en especies de hábitats cerrados (peces, aves y pequeños mamíferos)
4. Peluquería e higiene solo pesan en perros, gatos y conejos

Conclusiones específicas:
1. Fenómeno de veterinaria en  ```fish```: Tiene sentido ya que el gasto diario de mantenimiento de un pez, es muy económico, por lo que si el pez llega a necesitar una consulta
o tratamiento, su costo será obviamente alto y representará un gran porcentaje en el gasto anual

````python
# Porcentaje de animales asegurados por especie (Human made)
seguro_por_especie = (df.groupby("species")["has_pet_insurance"].mean() * 100).sort_values(ascending=False)
promedio_general = df["has_pet_insurance"].mean() * 100

fig, ax = plt.subplots(figsize=(10, 5.5))
# Paleta Set2 ajustada a la cantidad de especies
palette_seguro = sns.color_palette("Set2", n_colors=len(seguro_por_especie))
# Gráfico de barras
bars = ax.bar(
    seguro_por_especie.index,
    seguro_por_especie.values,
    color=palette_seguro,
    edgecolor="black",
    linewidth=0.8,
    width=0.55,
)
# Agregar etiquetas de porcentaje sobre cada barra
for bar in bars:
  altura = bar.get_height()
  ax.annotate(
      f"{altura:.1f}%",
      xy=(bar.get_x() + bar.get_width() / 2, altura),
      xytext=(0, 4),  # Desplazamiento hacia arriba
      textcoords="offset points",
      ha="center",
      va="bottom",
      fontsize=10.5,
      fontweight="bold",
  )
# Línea punteada que muestra el promedio general de la población
ax.axhline(
    promedio_general,
    color="#e74c3c",
    linestyle="--",
    linewidth=1.2,
    label=f"Promedio General ({promedio_general:.1f}%)",
)
# Títulos y estética
plt.title(
    "Porcentaje de Mascotas con Seguro Médico por Especie",
    fontsize=13,
    fontweight="bold",
    pad=15,
)
plt.xlabel("Especie", fontsize=11, fontweight="bold", labelpad=10)
plt.ylabel(
    "Tasa de Aseguramiento (%)", fontsize=11, fontweight="bold", labelpad=10
)
plt.ylim(0, max(seguro_por_especie.values) + 8)  # Espacio para las etiquetas
# Formato de porcentaje en el eje Y
ax.yaxis.set_major_formatter(ticker.PercentFormatter(decimals=0))

plt.xticks(fontsize=10.5)
plt.yticks(fontsize=10)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.legend(loc="upper right", frameon=True, fontsize=10)
sns.despine(top=True, right=True)

plt.tight_layout()
plt.show()
````

Conclusiones:
1. Ya que es observable el enorme gap entre perros-gatos y el resto de especies, podemos atribuirle esto principalmente a dos factores, naturaleza de las mascotas y disponibilidad en el mercado

2. Aunque especies como conejos o reptiles tienen costos de atención médica especializada similares o más altos a los de un perro o gato, la gran mayoría de los dueños no los tienen asegurados, lo que amplía notoriamente el costo médico para estas mascotas

**CONCLUSIÓN FINAL II**

* El seguro médico es el amortiguador financiero contra los gastos extremos: mientras que casi el 30% de los dueños de perros y gatos se protege mediante seguros contra los altos costos de enfermedades crónicas o accidentes, los dueños de animales pequeños y exóticos enfrentan una vulnerabilidad financiera total al asumir el costo médico de especialistas sin respaldo asegurado

````python
#Dataframe nuevo
# Seleccionar las variables numéricas más relevantes
columnas_analisis = [
    "age_years",
    "weight_kg",
    "household_size",
    "vet_visits_per_year",
    "daily_exercise_minutes",
    "monthly_food_cost_usd",
    "annual_vet_cost_usd",
    "total_annual_cost_usd",
    "owner_satisfaction_rating",
]
# Nombres legibles en español
etiquetas_espanol = [
    "Edad",
    "Peso (kg)",
    "Metros Hogar",
    "Visitas Vet / Año",
    "Ejercicio Diario",
    "Comida Mensual ($)",
    "Veterinaria Anual ($)",
    "Costo Anual Total ($)",
    "Satisfacción Dueño",
]
# Calcular correlación de Pearson (código asistido por IA para poder hacerlo)
matriz_corr = df[columnas_analisis].corr()
matriz_corr.columns = etiquetas_espanol
matriz_corr.index = etiquetas_espanol

fig, ax = plt.subplots(figsize=(10, 8))
# Gráfico de calor con escala divergente y valores anotados
sns.heatmap(
    matriz_corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",  # Azul = correlación negativa, Rojo = positiva, Blanco = neutro
    vmin=-1,
    vmax=1,
    center=0,
    linewidths=0.5,
    linecolor="white",
    cbar_kws={
        "label": "Coeficiente de Correlación de Pearson",
        "shrink": 0.8,
    },
    ax=ax,
)
plt.title(
    "Matriz de Correlación entre Variables Cuantitativas",
    fontsize=14,
    fontweight="bold",
    pad=15,
)
plt.xticks(rotation=45, ha="right", fontsize=10.5)
plt.yticks(rotation=0, fontsize=10.5)

plt.tight_layout()
plt.show()
````

¿Por qué se usa el coeficiente de Correlación de Pearson?
